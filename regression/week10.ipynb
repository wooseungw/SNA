```python
from sklearn.preprocessing import LabelEncoder

items = ["냉장고", "세탁기", "에어컨", "냉장고", "세탁기", "에어컨"]

encoder = LabelEncoder()
encoder.fit(items)

labels = encoder.transform(items)

print("Label Encoded Items:", labels)
print("Original Items:", items)
print("Classes:", encoder.classes_)
```

```python
from sklearn.preprocessing import MinMaxScaler
import numpy as np

train_array = np.arange(11).reshape(-1, 1)
print("Train Array:", train_array.reshape(-1))
scaler = MinMaxScaler()
scaler.fit(train_array)

train_scaled_array = scaler.transform(train_array)
print("Scaled Train Array:", train_scaled_array.reshape(-1))
print("="*50)
test_array = np.array([[0], [5], [10]])
print("Test Array:", test_array.reshape(-1))
scaled_test_array = scaler.transform(test_array)
print("Scaled Test Array:", scaled_test_array.reshape(-1))
```

```python
from sklearn.preprocessing import StandardScaler
from sklearn.datasets import load_iris
import pandas as pd

iris = load_iris()
iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)

print(iris_df.mean())
print(iris_df.std())

scaler = StandardScaler()
scaler.fit(iris_df)
scaled_iris = scaler.transform(iris_df)
scaled_iris_df = pd.DataFrame(data=scaled_iris, columns=iris.feature_names)
print(scaled_iris_df.mean())
print(scaled_iris_df.std())
```
