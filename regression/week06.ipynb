```python
data = {'id': ['1000', '1001', '1002', '1003', '1004'],
        'name': ['Alice', 'Bob', 'Charlie', 'David', 'Eva'],
}
df = pd.DataFrame(data)
data = [
    ['1000',  90.72],
    ['1001',  85.55],
    ['1002',  70.32],
    ['1003',  95.12],
    ['1004',  88.45]
]
df2 = pd.DataFrame(data, columns=['id', 'score'])
df2.set_index('id', inplace=True)
df2["grade"] = ['A', 'B', 'C', 'A', 'B']

df = pd.merge(df,df2, how='inner', on='id')
df.set_index('id', inplace=True)

df
```

```python
dic = {"이름": ["철수", "영희", "민철", "경수", "동수", "소연", "민수"],
"학번" : [1000, 1001, 1002,1003,1004,1005,1006],
"지역": ["서울", "대전", "대구","부산", "서울", "제주","서울"],
"국어": [84, 80, 81, 80, 97, 96, 99],
"영어": [81, 91, 97, 99, 96, 80, 85],
"수학": [87, 72, 86, 76, 79, 90, 95],
"과학": [74, 83, 75, 71, 98, 73, 80] }

df = pd.DataFrame(dic)
df.set_index("학번", inplace=True)
df["총합"] = df.loc[:,["국어", "영어", "수학", "과학"]].sum(axis=1)
df["평균"] = df.loc[:,["국어", "영어", "수학", "과학"]].mean(axis=1)
df
```
