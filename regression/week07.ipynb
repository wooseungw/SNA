```python
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.model_selection import train_test_split
import pandas as pd

iris = load_iris()

iris_df = pd.DataFrame(data=iris.data, columns=iris.feature_names)
print(iris.target_names)
iris_df['target'] = iris.target
iris_df.head()
```

```python
import joblib

x_t ,x_v, y_t, y_v = train_test_split(iris.data, iris.target, test_size=0.2, random_state=42)
clf = DecisionTreeClassifier(random_state=42)
clf.fit(x_t, y_t)
pred = clf.predict(x_v)
pred[:5]
# Save the model
joblib.dump(clf, 'decision_tree_model.pkl')

# Load the model
loaded_clf = joblib.load('decision_tree_model.pkl')
loaded_pred = loaded_clf.predict(x_v)
loaded_pred[:5]
```

```python
from sklearn.metrics import accuracy_score
accuracy = accuracy_score(y_v, pred)
print(f"Accuracy: {accuracy:.2f}")
```
