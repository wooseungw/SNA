```python
import numpy as np
import matplotlib.pyplot as plt

# 매개변수 t의 범위를 설정 (0부터 2π까지 1000개의 점)
t = np.linspace(0, 2 * np.pi, 1000)

# 하트 모양을 만드는 파라메트릭 방정식
x = 16 * np.sin(t)**3
y = 13 * np.cos(t) - 5 * np.cos(2*t) - 2 * np.cos(3*t) - np.cos(4*t)

# 그래프 출력
plt.figure(figsize=(6,6))
plt.plot(x, y, 'r')
plt.title("하트 모양 (파라메트릭 표현)")
plt.axis('equal')
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
