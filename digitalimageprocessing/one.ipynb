```python
import cv2

def main():
    # 웹캠을 열기 (Mac에서는 일반적으로 0번 장치 사용)
    cap = cv2.VideoCapture(index=0)
    
    # 웹캠이 제대로 열리지 않을 경우 처리
    if not cap.isOpened():
        print("웹캠을 열 수 없습니다.")
        return

    while True:
        # 웹캠에서 프레임 읽어오기
        ret, frame = cap.read()
        if not ret:
            print("프레임을 읽어올 수 없습니다.")
            break
        
        # 프레임을 그레이스케일로 변환 (Canny 알고리즘은 단일 채널 이미지에서 작동)
        gray = cv2.cvtColor(frame, cv2.COLOR_BGR2GRAY)
        
        # Canny Edge 검출 적용
        # threshold1, threshold2: 에지 검출 민감도 조절 파라미터
        edges = cv2.Canny(gray, threshold1=100, threshold2=200)
        
        # 원본 프레임과 에지 검출 결과를 화면에 출력
        cv2.imshow("원본", frame)
        cv2.imshow("Canny Edge", edges)
        
        # 'q' 키를 누르면 종료
        if cv2.waitKey(1) & 0xFF == ord('q'):
            break

    # 자원 해제
    cap.release()
    cv2.destroyAllWindows()

if __name__ == "__main__":
    main()
```
