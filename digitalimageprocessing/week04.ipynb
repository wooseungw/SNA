```python
import numpy as np
import cv2

switch_case = {
    ord('a'): "a 키 입력", ord('b'): "b 키 입력",
    0x41: "A 키 입력", 0x42: "B 키 입력",
}

image = np.ones((200, 300), np.uint8) * 255
cv2.namedWindow('Keyboard Event')
cv2.imshow('Keyboard Event', image)

while True:
    key = cv2.waitKeyEx(100)    # 100ms 동안 키 이벤트 대기
    if key == 27:               # ESC 키
        break
    try:
        result = switch_case[key]
        print(result)
    except KeyError:
        result = -1
cv2.destroyAllWindows()
```

마우스-키보드 인터렉티브를 통한 필터 변경 및 강도 조정

```python
import numpy as np
import cv2

def onMouse(event,x,y,flags,param):
    if event == cv2.EVENT_LBUTTONDOWN:
        print("마우스 왼쪽 버튼 누르기")
    elif event == cv2.EVENT_RBUTTONDOWN:
        print("마우스 오른쪽 버튼 누르기")
    elif event == cv2.EVENT_LBUTTONUP:
        print("마우스 왼쪽 버튼 떼기")
    elif event == cv2.EVENT_RBUTTONUP:
        print("마우스 오른쪽 버튼 떼기")
    elif event == cv2.EVENT_LBUTTONDBLCLK:
        print("마우스 왼쪽 버튼 더블클릭")
    elif event == cv2.EVENT_RBUTTONDBLCLK:
        print("마우스 오른쪽 버튼 더블클릭")
    elif event == cv2.EVENT_MOUSEMOVE:
        print("마우스 움직임:", x, y)
        
        
image = np.full((200, 300), 255, np.uint8)    # 200x300 행렬, 255(흰색)로 채움

title = 'Mouse Event'                          # 윈도우 이름
cv2.imshow(title, image)
cv2.setMouseCallback(title, onMouse)           # 마우스 이벤트 등록
cv2.waitKey(0)
cv2.destroyAllWindows()
```

```python
import numpy as np
import cv2

orange, blue, cyan = (0, 165, 255), (255, 0, 0), (255, 255, 0)
white, black = (255, 255, 255), (0, 0, 0)
image = np.full((300, 500, 3), white, np.uint8)    # 300x500, 흰색 영상 생성

pt1, pt2 = (120, 50), (380, 220)    # 좌상단, 우하단 좌표
pt3, pt4 = (100, 200), (30, 250)    # 좌상단, 우하단 좌표
roi = (90, 50, 90, 170)             # 좌상단 x, y, 너비, 높이

cv2.rectangle(image, pt1, pt2, blue, 3)
cv2.line(image, pt1, pt2, black)

cv2.line(image, pt3, pt4, cyan, 3, cv2.LINE_AA)
cv2.rectangle(image, pt3,pt4, cyan, 3)

cv2.rectangle(image, roi, orange, 3)

cv2.imshow("Line & Rectangle", image)
cv2.waitKey(0)
cv2.destroyAllWindows()
```
